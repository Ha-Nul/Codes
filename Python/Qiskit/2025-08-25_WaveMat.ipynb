```python
import numpy as np
import matplotlib.pyplot as plt
import qiskit

from qiskit import QuantumCircuit, QuantumRegister
from qiskit_aer import AerSimulator
```

## Circuit
Run This code to reset the circuit

```python
q_s = QuantumRegister(1, name='S')
q_r = QuantumRegister(1, name='R')
q_H = QuantumRegister(1, name='H')
q_P = QuantumRegister(1, name='P')
q_Q = QuantumRegister(1, name='Q')

qc = QuantumCircuit(q_s,q_r,q_H,q_P,q_Q)
```

### Gates (Operations)

```python
qc.swap(q_s,q_H)
qc.barrier
qc.h(q_P[0])
qc.cx(q_P[0],q_Q[0])
```

```python
qc.draw('mpl',style='iqx')
```

#### Local simulator Run

```python
# 시뮬레이터 준비
simulator = AerSimulator()

# 실행
job = simulator.run(qc)
result = job.result()
counts = result.get_counts()
```
