```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
targ_arr = np.random.randint(1,10,size=10) #임의의 에너지값 배열
```

```python
prob_arr = np.arange(10)
arr_prob = np.sort(prob_arr)[::-1] #내림차순(nonincreasing order) 확률 배열

arr_targ = np.sort(targ_arr)[::]
ttarg_arr = np.sort(arr_targ)[::-1]
ran_arr = np.random.permutation(targ_arr)

case1 = np.zeros(10)
case2 = np.zeros(10)
case3 = np.zeros(10)
case4 = np.zeros(10)

subtract1 = np.zeros(10)
subtract2 = np.zeros(10)
subtract3 = np.zeros(10)

for i in range(10):
    case1[i] = arr_prob[i] * targ_arr[i] # 확률은 내림차순, 무작위로 섞인 곱
    case2[i] = arr_prob[i] * arr_targ[i] # 확률은 내림차순, 곱은 올림차순
    case3[i] = arr_prob[i] * ttarg_arr[i] # 확률은 내림차순, 곱은 내림차순
    case4[i] = arr_prob[i] * ran_arr[i] # 확률은 내림차순, 곱은 무작위에 permutation

    subtract1[i] = case1[i] - case2[i]
    subtract2[i] = case1[i] - case3[i]
    subtract3[i] = case1[i] - case4[i]

var1 = np.sum(case2)
var2 = np.sum(case3)
var3 = np.sum(case4)
```

```python
ran_arr = np.random.permutation(arr_targ)
print(ran_arr)
```

```python
plt.plot(prob_arr,case1)
plt.plot(prob_arr,case2)
plt.plot(prob_arr,case3)

plt.plot(prob_arr,subtract1,linestyle='--',linewidth=1)
plt.plot(prob_arr,subtract2,linestyle='--',linewidth=1)
plt.plot(prob_arr,subtract3,linestyle='--',linewidth=1)

print(var1,var2,var3)
print(ran_arr)
```
