```python
import numpy as np
import scipy as scipy

import matplotlib.pyplot as plt
np.set_printoptions(legacy='1.13',linewidth=np.inf, threshold=np.inf, precision=3)
```

```python
omega = 2
temp = np.linspace(1,100,200)

ham_dist = lambda m, om, t : np.exp(-m*om/t)*(1-np.exp(-om/t))

ham_x = [np.exp(-2/25),np.exp(-2/50),np.exp(-2/100)]

therm = [0,ham_x[0],ham_x[0]+ham_x[1],sum(ham_x)]
```

```python
dist1 = [ham_dist(1,1,25),ham_dist(1,1,50),ham_dist(1,1,100)]
dist2 = [ham_dist(0.2,1,25),ham_dist(0.2,1,50),ham_dist(0.2,1,100)]

sort1 = [0]
sort2 = [0]

sort1.extend(sorted(dist1))
sort2.extend(sorted(dist2))

sort1 = np.array(sort1)
sort2 = np.array(sort2)

sort1 = sort1/np.linalg.norm(sort1)
sort2 = sort2/np.linalg.norm(sort2)

print("sort1: ",sort1)
```

```python
plt.plot(therm,sort1)
plt.plot(therm,sort2)
```
