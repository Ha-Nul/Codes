```python
import numpy as np
import scipy as sp

import matplotlib.pyplot as plt
np.set_printoptions(legacy='1.13',linewidth=np.inf, threshold=np.inf, precision=3)
```

```python
#hbar = 1

mass = 1
omega = np.ones(10)
t_grid = np.linspace(0,2,20)

p_omega = [i*omega[i] for i in range(len(omega))]
```

```python
t = 1
U_matrix = np.zeros((len(omega),len(omega)),dtype=complex)

for i in range(len(omega)):
    for j in range(len(omega)):
        if i==j:
            U_matrix[i][j] = np.exp((i+1/2)*omega[i]*-1j*t)
        else:
            None
```

```python
p_matrix = np.zeros((len(omega),len(omega)),dtype=complex)

for i in range(len(omega)):
    for j in range(len(omega)):
        if j == i+1:
            p_matrix[i][j] = 1j *np.sqrt(mass*p_omega[i+1]/2)
        elif j == i-1 and i != 0:
            p_matrix[i][j] = -1j *np.sqrt(mass*p_omega[i]/2)
```


Dephasing channel trial (Not complete dephasing channel)

```python
dephas = np.zeros((len(omega), len(omega)))
prob_dephas = 0.5

# 작용 자체가 행렬로 나타내어지는 것이 아니라, 어떤 함수로써 들어온 상태의 행렬 원소에 연산이 가해지므로 함수의 형태로 작성해야 함.
def deph(dephas,prob_dephas):
    for i in range(len(omega)):
        for j in range(len(omega)):
            if i==j:
                dephas[i][j] = 1
            elif i != j:
            #elif i == j+1 or i == j-1:
                #dephas[i][j] = prob_dephas
                dephas[i][j] = 0 #complete dephasing channel
    return dephas
```

제미나이 답지

```python
def partial_dephasing_channel(rho_input, gamma):
    """
    입력 밀도 행렬 rho_input의 비대각 성분에 gamma를 곱합니다.
    대각 성분은 그대로 둡니다.
    """
    rho_dephased = rho_input.copy() # 원본 유지를 위해 복사
    for m in range(rho_input.shape[0]):
        for n in range(rho_input.shape[1]):
            if m != n:
                rho_dephased[m, n] *= gamma
    return rho_dephased

# 사용 예시
# rho_initial = ... (초기 밀도 행렬)
# gamma_factor = 0.5 
# rho_after_1_step = partial_dephasing_channel(rho_initial, gamma_factor)
# rho_after_2_steps = partial_dephasing_channel(rho_after_1_step, gamma_factor) 
# ...
```

제미나이가 만든 답지 코드

```python
import numpy as np
import scipy.linalg

# --- 물리 상수 및 파라미터 (이전 코드에서 가져옴) ---
hbar = 1.0
m_particle = 1.0
omega_0 = 1.0    # 조화 진동자 각진동수 (스칼라)
N_dim = 10       # 힐베르트 공간 차원
t = 1.0          # 특정 시간 (U(t)에 사용되지만, A^(k) 추출 자체에는 불필요)

# --- 해밀토니안 H_QHO 및 시간 변화 연산자 U(t) (모드 추출에는 U(t) 직접 불필요) ---
# H_qho_diag = np.array([(n + 0.5) * hbar * omega_0 for n in range(N_dim)])
# U_t_diag = np.exp(-1j * H_qho_diag * t / hbar)
# U_t_matrix = np.diag(U_t_diag)

# --- 운동량 연산자 p_matrix (A로 사용될 연산자) ---
C_p = 1j / np.sqrt(2.0) # m=hbar=omega_0=1 가정 시의 상수 (이전 코드와 동일)
A_operator = np.zeros((N_dim, N_dim), dtype=complex) # p_matrix_corrected 대신 A_operator로 명명
for r_idx in range(N_dim): # row index (m)
    for c_idx in range(N_dim): # col index (n)
        if r_idx == c_idx + 1: # <m|p|n> where m = n+1
            A_operator[r_idx, c_idx] = C_p * np.sqrt(c_idx + 1) # c_idx = n
        elif r_idx == c_idx - 1: # <m|p|n> where m = n-1
             if c_idx > 0: # n > 0
                A_operator[r_idx, c_idx] = -C_p * np.sqrt(c_idx)  # c_idx = n
print("Operator A (Momentum Operator p):")
print(A_operator)
print("-" * 30)

# --- 모드 분리 ---
# k_prime 은 m - n 값을 의미 (에너지 차이에 비례)
# k_prime의 범위: -(N_dim-1) 부터 (N_dim-1) 까지

A_modes = {} # 각 모드 성분 행렬을 저장할 딕셔너리

for k_prime in range(-(N_dim - 1), N_dim):
    A_mode_k = np.zeros_like(A_operator, dtype=complex) # 현재 모드 성분을 저장할 행렬
    for m_idx in range(N_dim): # row index
        for n_idx in range(N_dim): # col index
            if (m_idx - n_idx) == k_prime:
                A_mode_k[m_idx, n_idx] = A_operator[m_idx, n_idx]
    
    # 의미있는 모드(0이 아닌 행렬)만 저장하거나, 모두 저장할 수 있음
    if np.any(A_mode_k): # A_mode_k 행렬에 0이 아닌 요소가 하나라도 있다면
        A_modes[k_prime] = A_mode_k

# --- 결과 확인 ---
print("Decomposed Asymmetry Modes A^(k):")
for k_val, mode_matrix in A_modes.items():
    print(f"Mode k = {k_val}:")
    print(mode_matrix)
    # 각 모드의 크기 (예: Frobenius norm) 계산 및 시각화 가능
    # norm_k = np.linalg.norm(mode_matrix)
    # print(f"Norm of mode k={k_val}: {norm_k:.3f}")
    print("-" * 10)

# --- 운동량 연산자의 경우 예상 결과 ---
# 운동량 연산자는 <m|p|n>에서 m = n+1 또는 m = n-1 일 때만 0이 아닌 값을 가집니다.
# 따라서 k_prime = m-n = +1 (첫 번째 부대각선) 또는 
# k_prime = m-n = -1 (첫 번째 초 대각선) 인 모드만 0이 아닌 성분을 갖게 됩니다.
# A_modes 딕셔너리에는 k_val이 1과 -1인 두 개의 키만 존재해야 합니다.

# 예시: k=1 모드 (A_modes[1])는 A_operator의 첫 번째 부대각선 성분만 가짐
# 예시: k=-1 모드 (A_modes[-1])는 A_operator의 첫 번째 초 대각선 성분만 가짐

# --- "비대칭성 스펙트럼" 시각화 아이디어 ---
mode_indices = sorted(A_modes.keys())
mode_norms = [np.linalg.norm(A_modes[k]) for k in mode_indices]

# import matplotlib.pyplot as plt
plt.figure(figsize=(8, 5))
plt.bar(mode_indices, mode_norms)
plt.xlabel("Mode index k (m-n)")
plt.ylabel("Norm of A^(k)")
plt.title("Asymmetry Mode Spectrum of Operator A (Momentum Operator)")
plt.xticks(mode_indices)
plt.grid(True)
plt.show()
```
