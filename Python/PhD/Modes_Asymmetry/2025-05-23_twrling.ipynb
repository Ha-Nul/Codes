```python
import numpy as np
import scipy as sp

import matplotlib.pyplot as plt
np.set_printoptions(legacy='1.13',linewidth=np.inf, threshold=np.inf, precision=2)
```

```python
dim = 10
```

```python
high_dim_X = np.zeros((dim,dim))
high_dim_Z = np.zeros((dim,dim), dtype=complex)

#Qudit Pauli X gate
for i in range(dim):
    for j in range(dim):
        if i==0 and j==dim-1:
            high_dim_X[i][j] = 1
        elif j == i-1 and i != 0:
            high_dim_X[i][j] = 1

for i in range(dim):
    for j in range(dim):
        if i==j:
            high_dim_Z[i][j] = np.exp(1j * (2*np.pi/dim) * i)
```

```python
high_dim_Z
```

```python
Unit = (high_dim_X @ high_dim_Z)/(dim**2)
```

```python
Unit
```

```python
plt.imshow(Unit.real)
plt.colorbar()
```
